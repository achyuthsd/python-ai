```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("students.csv")
city_order = sorted(df["City"].dropna().unique())

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(data=df, x="Math_Score", bins=5, ax=axes[0])
axes[0].set_title("Distribution of Math scores")

sns.scatterplot(data=df, x="Science_Score", y="Math_Score", ax=axes[1])
axes[1].set_title("Science and Math scores")

sns.boxplot(data=df, x="City", y="Math_Score", order=city_order, ax=axes[2])
axes[2].set_title("Math score spread by city")
axes[2].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("students.csv")

club_counts = df["Club"].fillna("Not recorded").value_counts().sort_index()
number_columns = ["Math_Score", "Science_Score", "Attendance_Percent"]
correlations = df[number_columns].corr()

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].pie(club_counts.values, labels=club_counts.index, autopct="%1.0f%%")
axes[0].set_title("Student club choices")

sns.heatmap(correlations, annot=True, fmt=".2f", vmin=-1, vmax=1, ax=axes[1])
axes[1].set_title("Correlation between number columns")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("students.csv")
complete_data = df.dropna(subset=["Science_Score", "Math_Score"])

X = complete_data[["Science_Score"]]
y = complete_data["Math_Score"]
names = complete_data["Name"]

X_train, X_test, y_train, y_test, names_train, names_test = train_test_split(
    X, y, names, test_size=0.25, random_state=42
)

print("Complete student examples:", len(complete_data))
print("Feature:", list(X.columns))
print("Target:", y.name)
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Training students:", names_train.tolist())
print("Test students:", names_test.tolist())
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


df = pd.read_csv('studentperf.csv')
print('--- First Five Rows ---')
print(df.head())
print('\n--- Shape (Rows, Columns) ---')
print(df.shape)
print('\n--- Column Names & Data Types ---')
print(df.dtypes)
print('\n--- Missing Value Count ---')
print(df.isnull().sum())
print('\n--- Duplicate Count ---')
print(df.duplicated().sum())
```

```python
num_cols = ['StudyTimeWeekly', 'GPA']
print('\n--- NumPy Statistics ---')
for col in num_cols:
  arr = df[col].to_numpy()
  print(
      f'{col} -> Mean: {np.mean(arr):.4f}, Min: {np.min(arr):.4f}, Max:'
      f' {np.max(arr):.4f}'
  )

print('\n--- Pandas Summary Statistics ---')
print(df[num_cols].describe())

sorted_df = df.sort_values(by='GPA', ascending=False)
print('\n--- First Five Rows After Sorting by GPA (Descending) ---')
print(sorted_df.head())
```

```python
for col in ['StudyTimeWeekly', 'Absences', 'GPA']:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  outliers = df[(df[col] < q1 - 1.5 * iqr) | (df[col] > q3 + 1.5 * iqr)]
  print(f'Outliers in {col}: {len(outliers)}')
```

```python
print('\n--- Descriptive Statistics ---')
for col in num_cols:
  arr = df[col].to_numpy()
  print(
      f'{col} -> Median: {np.median(arr):.4f}, Variance:'
      f' {np.var(arr, ddof=1):.4f}, Standard Deviation:'
      f' {np.std(arr, ddof=1):.4f}'
  )

print('\n--- Correlation Matrix ---')
print(df[num_cols].corr())
```
